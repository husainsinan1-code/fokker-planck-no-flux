## Setting up

```python
import numpy as np
import matplotlib.pyplot as plt

import os
```

```python
import sys
sys.path.append("..")
```

```python
from sources.simulation import numerical_solution, euler_maruyama
from sources.analysis import stationary_solution
from sources.models import *
from sources.data_storage import *
```

```python

#defind varaibles

#10 mil
n_parts = 10000000

T = 2

# dt_list = [0.000625]#, 0.00015625]
dt_list = [0.00015625]

#set the boundries
x_bounds = (0, 1)


X_0 = np.random.uniform(.45,.55, n_parts)

theta_list = [0.25, 0.5, 0.75]

boundaries = ["reflection", "rejection"]#, "theta", "half_step"]

```

## Case $b=0$ and $\sigma = 1$

```python
b = lambda x, t=None: constant_drift(x, value=0)
sigma = lambda x, t=None: constant_diffusion(x, value=1)

path = "./data/b-0_s-1"

for dt in dt_list:

    n_steps = int(T/dt)

    for boundary in boundaries:
        if boundary == "theta":
            for theta in theta_list:
                X = numerical_solution(method=euler_maruyama, X_0=X_0, dt=dt, n_steps=n_steps
                                                     , b=b, sigma=sigma, boundary_type=(boundary, theta)
                                                     , boundaries=x_bounds)
                # filename = boundary + "-" + theta + "_dt-" + dt
                filename = (f"{boundary}-{theta}_dt-{dt}")
                save_solution(X, filename, path)

        else:
            X = numerical_solution(method=euler_maruyama, X_0=X_0, dt=dt, n_steps=n_steps
                                                                 , b=b, sigma=sigma, boundary_type=boundary
                                                                 , boundaries=x_bounds)
            filename = (f"{boundary}_dt-{dt}")
            save_solution(X, filename, path)

    
    print(f"finished {dt=}")


del sigma, b
```
