```python
import numpy as np
import matplotlib.pyplot as plt

import sys
sys.path.append("..")

from sources.simulation import apply_boundary_2d, get_boundary_normal
```

```python

bounds = ((0, 1), (0, 1))

sigma_normal = np.eye(2)

sigma_oblique = np.array([[1.0, 0.5],
                          [0.5, 1.0]])
```

# Case crossing the right boundary only

```python
X_current = np.array([[0.5, 0.5]])
proposal = np.array([[1.2, 0.7]])

normal, boundary_value, reflection_point = get_boundary_normal(X_current, proposal, bounds)

v_normal = normal @ sigma_normal.T
v_normal = v_normal/np.linalg.norm(v_normal, axis=1)[:, None]

v_oblique = normal @ sigma_oblique.T
v_oblique = v_oblique/np.linalg.norm(v_oblique, axis=1)[:, None]

X_final_normal = apply_boundary_2d(proposal.copy(), X_current.copy(), "reflection", bounds, sigma_normal)
X_final_oblique = apply_boundary_2d(proposal.copy(), X_current.copy(), "reflection", bounds, sigma_oblique)


fig, ax = plt.subplots(1, 2, figsize=(12, 5), sharex=True, sharey=True)

ax[0].plot([0, 1, 1, 0, 0], [0, 0, 1, 1, 0])
ax[0].scatter(X_current[:, 0], X_current[:, 1], label="current")
ax[0].scatter(proposal[:, 0], proposal[:, 1], label="proposal")
ax[0].scatter(reflection_point[:, 0], reflection_point[:, 1], marker="x", s=100, label="reflection point")
ax[0].scatter(X_final_normal[:, 0], X_final_normal[:, 1], label="final reflected")
ax[0].plot([X_current[0, 0], proposal[0, 0]], [X_current[0, 1], proposal[0, 1]], "--", label="attempt")
ax[0].plot([proposal[0, 0], X_final_normal[0, 0]], [proposal[0, 1], X_final_normal[0, 1]], label="reflection move")
ax[0].arrow(reflection_point[0, 0], reflection_point[0, 1], 0.16*v_normal[0, 0], 0.16*v_normal[0, 1], head_width=0.02, length_includes_head=True)
ax[0].set_title("normal reflection")

ax[1].plot([0, 1, 1, 0, 0], [0, 0, 1, 1, 0])
ax[1].scatter(X_current[:, 0], X_current[:, 1], label="current")
ax[1].scatter(proposal[:, 0], proposal[:, 1], label="proposal")
ax[1].scatter(reflection_point[:, 0], reflection_point[:, 1], marker="x", s=100, label="reflection point")
ax[1].scatter(X_final_oblique[:, 0], X_final_oblique[:, 1], label="final reflected")
ax[1].plot([X_current[0, 0], proposal[0, 0]], [X_current[0, 1], proposal[0, 1]], "--", label="attempt")
ax[1].plot([proposal[0, 0], X_final_oblique[0, 0]], [proposal[0, 1], X_final_oblique[0, 1]], label="reflection move")
ax[1].arrow(reflection_point[0, 0], reflection_point[0, 1], 0.16*v_oblique[0, 0], 0.16*v_oblique[0, 1], head_width=0.02, length_includes_head=True)
ax[1].set_title("oblique reflection")

for a in ax:
    a.set_xlim(-0.2, 1.4)
    a.set_ylim(-0.2, 1.4)
    a.set_aspect("equal")
    a.grid(True)

handles, labels = ax[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="lower center", bbox_to_anchor=(0.5, -0.05), ncol=6)
plt.tight_layout()
plt.show()
```
