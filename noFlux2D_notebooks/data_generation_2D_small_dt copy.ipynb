## Parameters

```python
n_parts = 10000000
# n_parts = 100000

T = 2
# dt_list = [0.16, 0.04, 0.01]
dt_list =[0.0025]

radius = 1.0
boundaries = ("disk", radius)

X_0 = initialize_disk(n_parts)

boundary_types = ["reflection", "rejection", "reflect_norm"]
```

## Case $b=(0,0)$ and $\sigma=I$

```python
b = np.array([0.0, 0.0])
sigma = np.eye(2)

path = "./data/disk_b-0-0_s-I"

for dt in dt_list:

    n_steps = int(T/dt)

    for boundary_type in boundary_types:

        X = numerical_solution(method=euler_maruyama_2d, X_0=X_0, dt=dt, n_steps=n_steps,
                               b=b, sigma=sigma, boundary_type=boundary_type, boundaries=boundaries)

        filename = f"{boundary_type}_dt-{dt}"
        save_solution(X, filename, path)

    print(f"finished {dt=}")

del b, sigma
```

## Case $b=(0,1)$ and $\sigma=I$

```python
b = np.array([0.0, 1.0])
sigma = np.eye(2)

path = "./data/disk_b-0-1_s-I"

for dt in dt_list:

    n_steps = int(T/dt)

    for boundary_type in boundary_types:

        X = numerical_solution(method=euler_maruyama_2d, X_0=X_0, dt=dt, n_steps=n_steps,
                               b=b, sigma=sigma, boundary_type=boundary_type, boundaries=boundaries)

        filename = f"{boundary_type}_dt-{dt}"
        save_solution(X, filename, path)

    print(f"finished {dt=}")

del b, sigma
```
