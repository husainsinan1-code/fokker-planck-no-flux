## Setting up

```python
import numpy as np
import matplotlib.pyplot as plt

import os
```

```python
from sources.simulation import numerical_solution, euler_maruyama
from sources.plotting import *
from sources.analysis import stationary_solution
from sources.models import *
from sources.data_storage import *
```

```python

#defind varaibles

#10 mil
n_parts = 10000000

T = 2

dt_list = [0.04, 0.01, 0.0025, 0.000625]

# n_steps_list = [int(T/dt) for dt in dt_list]

# epsilon_list = [np.sqrt(dt) for dt in dt_list]

#set the boundries
x_bounds = (0, 1)

#drift and diffusion as functions

b = lambda x, t=None: constant_drift(x, t, value=1)
sigma = lambda x, t=None: constant_diffusion(x, t, value=1)



X_0 = np.random.uniform(.45,.55, n_parts)

```

## Generate Reflection boundary data

```python
for dt in dt_list:
    n_steps = int(T/dt)

    numerical_solution(method=euler_maruyama, X_0=X_0, dt=dt, n_steps=n_steps
                                                     , b=b, sigma=sigma, boundary_type="reflection"
                                                     , boundaries=x_bounds)
    
    print(f"finished saving {dt=}")
```

## Generate Rejection boundary data

```python
for dt in dt_list:
    n_steps = int(T/dt)

    numerical_solution(method=euler_maruyama, X_0=X_0, dt=dt, n_steps=n_steps
                                                     , b=b, sigma=sigma, boundary_type="rejection"
                                                     , boundaries=x_bounds)
    
    print(f"finished saving {dt=}")
```

## Generate $\theta$ boundary data for $\theta$ in `[0.25, 0.5, 0.75]`

```python
for theta in [0.025, 0.5, 0.75]:
    for dt in dt_list:
        n_steps = int(T/dt)

        numerical_solution(method=euler_maruyama, X_0=X_0, dt=dt, n_steps=n_steps
                                                        , b=b, sigma=sigma, boundary_type=("theta", theta)
                                                        , boundaries=x_bounds)
        
        print(f"finished saving {dt=}")
```
